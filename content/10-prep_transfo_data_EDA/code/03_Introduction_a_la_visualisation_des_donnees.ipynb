# __Introduction à la visualisation des données__

La visualisation des données est la représentation graphique de données dans le but de mettre en évidence des **tendances**, des **relations**, des **structures** ou des **informations importantes** qui ne sont pas toujours faciles à repérer dans les données brutes.

- Elle consiste à utiliser des éléments visuels comme des graphiques, des diagrammes et des cartes afin de communiquer des informations complexes sous une forme plus facile à comprendre et à interpréter.

![link text](https://labcontent.simplicdn.net/data-content/content-assets/Data_and_AI/ADSP_Images/Lesson_05_Data_Visualization/1_Introduction_to_Data_Visualization/Introduction_to_Data_Visualization.png)

Les outils et bibliothèques de visualisation, comme **Matplotlib**, **Seaborn** et **Plotly**, permettent aux analystes, aux scientifiques des données et aux professionnels de créer des représentations visuelles qui facilitent la compréhension des données et soutiennent la prise de décision fondée sur les données.

## __Graphique linéaire__

Un graphique linéaire représente une série de points reliés par des segments de droite.

Il est particulièrement utile pour :

- représenter l'évolution d'une variable dans le temps ;
- observer une tendance ;
- visualiser des variations sur un intervalle continu.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour un graphique linéaire
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Créer le graphique linéaire
plt.plot(x, y, label='Onde sinusoïdale', color='blue', linestyle='-', linewidth=2)

# Ajouter les étiquettes et le titre
plt.xlabel('Axe X')
plt.ylabel('Axe Y')
plt.title('Graphique linéaire : onde sinusoïdale')

# Ajouter une légende
plt.legend()

# Afficher le graphique
plt.show()
```

## __Nuage de points__

Un nuage de points (*scatter plot*) permet de représenter la relation entre **deux variables numériques continues**.

- Chaque point représente une observation.
- Sa position dépend des valeurs des deux variables.
- Ce graphique permet notamment de repérer des relations, des regroupements, des tendances ou des valeurs atypiques.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour un nuage de points
x = np.random.rand(100)
y = 2 * x + np.random.randn(100)

# Créer le nuage de points
plt.scatter(x, y, color='green', marker='o', label='Données aléatoires')

# Ajouter les étiquettes et le titre
plt.xlabel('Axe X')
plt.ylabel('Axe Y')
plt.title('Nuage de points : données aléatoires')

# Ajouter une légende
plt.legend()

# Afficher le graphique
plt.show()
```

## __Diagramme à barres__

Un diagramme à barres est fréquemment utilisé pour représenter des **données catégorielles**.

- Chaque catégorie est représentée par une barre.
- La longueur ou la hauteur de la barre est proportionnelle à la valeur représentée.
- Ce graphique facilite la comparaison entre plusieurs catégories.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour un diagramme à barres
categories = ['Catégorie A', 'Catégorie B', 'Catégorie C']
values = [25, 40, 15]

# Créer le diagramme à barres
plt.bar(categories, values, color='orange', edgecolor='black', label='Diagramme à barres')

# Ajouter les étiquettes et le titre
plt.xlabel('Catégories')
plt.ylabel('Valeurs')
plt.title('Diagramme à barres : comparaison des catégories')

# Ajouter une légende
plt.legend()

# Afficher le graphique
plt.show()
```

## __Boîte à moustaches__

Une boîte à moustaches (*box plot* ou *box-and-whisker plot*) fournit un résumé visuel de la distribution d'un jeu de données.

Elle permet notamment de représenter :

- la valeur minimale ;
- le premier quartile ;
- la médiane ;
- le troisième quartile ;
- la valeur maximale ;
- les valeurs aberrantes éventuelles.

Ce graphique est particulièrement utile pour comparer la **dispersion** et la **tendance centrale** de plusieurs groupes.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour une boîte à moustaches
data = [np.random.normal(0, 1, 100), np.random.normal(0, 1.5, 100), np.random.normal(0, 2, 100)]

# Créer la boîte à moustaches
plt.boxplot(data, labels=['Groupe 1', 'Groupe 2', 'Groupe 3'], patch_artist=True, notch=True)

# Ajouter les étiquettes et le titre
plt.xlabel('Groupes')
plt.ylabel('Valeurs')
plt.title('Boîte à moustaches : comparaison des groupes')

# Afficher le graphique
plt.show()
```

## __Diagramme radar__

Un diagramme radar, aussi appelé **diagramme en toile d'araignée**, permet de représenter plusieurs variables quantitatives sur des axes qui partent tous du même point.

Il est utile pour comparer le profil de plusieurs mesures ou caractéristiques.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer des données aléatoires pour un diagramme radar
categories = ['Catégorie A', 'Catégorie B', 'Catégorie C', 'Catégorie D', 'Catégorie E']
data = np.random.randint(1, 10, len(categories))

# Créer le diagramme radar
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False)
data = np.concatenate((data, [data[0]]))
angles = np.concatenate((angles, [angles[0]]))
plt.polar(angles, data, marker='o')

# Ajouter les étiquettes
plt.thetagrids(np.degrees(angles[:-1]), labels=categories)

# Afficher le graphique
plt.title('Diagramme radar : données aléatoires')
plt.show()
```

## __Graphique en aires__

Un graphique en aires représente des valeurs sur un intervalle continu en remplissant la zone située entre des courbes ou entre une courbe et un axe.

- Il permet de mettre en évidence l'amplitude ou l'évolution d'une quantité.
- Il peut aussi servir à illustrer la contribution de plusieurs composantes à un total.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour un graphique en aires
x = np.linspace(0, 5, 100)
y1 = x
y2 = x**2

# Créer le graphique en aires
plt.fill_between(x, y1, y2, alpha=0.5, label='Aire entre les courbes')

# Ajouter les étiquettes et le titre
plt.xlabel('Axe X')
plt.ylabel('Axe Y')
plt.title('Graphique en aires : zone entre deux courbes')

# Ajouter une légende
plt.legend()

# Afficher le graphique
plt.show()
```

## __Graphique polaire__

Un graphique polaire représente les données à l'aide de **coordonnées polaires** plutôt que de coordonnées cartésiennes classiques.

- Il est particulièrement utile pour visualiser des phénomènes cycliques, périodiques ou dépendant d'un angle.

```python
import matplotlib.pyplot as plt
import numpy as np

# Générer les données pour un graphique polaire
theta = np.linspace(0, 2 * np.pi, 100)
r = theta * 2

# Créer le graphique polaire
plt.polar(theta, r, label='Graphique polaire')

# Ajouter une légende
plt.legend()

# Afficher le graphique
plt.show()
```

## __Treemap__

Une **treemap** permet de représenter des données hiérarchiques à l'aide de rectangles imbriqués.

- La taille de chaque rectangle peut représenter une valeur quantitative.
- Elle donne une vue d'ensemble de la hiérarchie.
- Elle facilite la comparaison de l'importance relative des différentes branches ou catégories.

```python
#!pip install squarify
import matplotlib.pyplot as plt
import squarify

# Exemple de données hiérarchiques
data = {
    'Racine': {
        'Branche1': {'Feuille1': 20, 'Feuille2': 30},
        'Branche2': {'Feuille3': 25, 'Feuille4': 15}
    }
}

# Fonction pour aplatir les données hiérarchiques
def flatten_hierarchy(node, parent=''):
    items = []
    for key, value in node.items():
        if isinstance(value, dict):
            items.extend(flatten_hierarchy(value, parent=f'{parent}\n{key}' if parent else key))
        else:
            items.append((parent, key, value))
    return items

# Aplatir les données hiérarchiques pour créer la treemap
flat_data = flatten_hierarchy(data)

# Créer la treemap
fig, ax = plt.subplots(figsize=(6, 6))
squarify.plot(sizes=[item[2] for item in flat_data], label=[f'{item[0]}\n{item[1]}' for item in flat_data], alpha=0.7, ax=ax)

# Masquer les axes pour améliorer la lisibilité
plt.axis('off')

# Ajouter un titre
plt.title('Treemap : exemple de hiérarchie')

# Afficher le graphique
plt.show()
```

## __Diagramme circulaire__

Un diagramme circulaire (*pie chart*) est divisé en secteurs afin de représenter des proportions.

- Chaque secteur représente une partie du total.
- Le cercle complet correspond à **100 %**.
- Ce type de graphique peut être utilisé pour représenter la répartition d'une variable catégorielle comportant un petit nombre de catégories.

```python
import matplotlib.pyplot as plt

# Exemple de données
values = [30, 40, 30]
labels = ['Catégorie A', 'Catégorie B', 'Catégorie C']

# Créer le diagramme circulaire
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90)

# Ajouter un titre
plt.title('Exemple de diagramme circulaire')

# Afficher le graphique
plt.show()
```

## __Visualisation 3D avec Matplotlib__

Matplotlib permet également de créer des graphiques en **trois dimensions** afin de représenter simultanément plusieurs variables.

### __Principales fonctionnalités__

- **Création d'axes 3D :** Matplotlib permet de créer un système d'axes pour représenter des données dans trois dimensions.
- **Nuages de points, courbes et surfaces :** on peut créer des nuages de points, des courbes et des surfaces 3D afin d'illustrer des relations dans un espace tridimensionnel.
- **Personnalisation et étiquetage :** les axes, titres et éléments visuels peuvent être personnalisés afin de faciliter l'interprétation.
- **Intégration avec Matplotlib :** les graphiques 3D s'intègrent aux autres fonctionnalités de la bibliothèque Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Lire le fichier CSV dans un DataFrame Pandas
df = pd.read_csv('ADANIPORTS.csv', parse_dates=True)

# Calculer l'écart entre les prix maximum et minimum
df['H-L'] = df.High - df.Low

# Calculer la moyenne mobile sur 100 jours des prix de clôture
df['100MA'] = df['Close'].rolling(100).mean()

# Créer un nuage de points 3D
ax = plt.axes(projection='3d')
ax.scatter(df.index, df['H-L'], df['100MA'])

# Ajouter les étiquettes des axes
ax.set_xlabel('Index')
ax.set_ylabel('H-L')
ax.set_zlabel('100MA')

# Afficher le nuage de points 3D
plt.show()
```

```python
# Créer un système d'axes 3D
ax = plt.axes(projection='3d')

# Nuage de points 3D avec l'index, 'H-L' et le volume
ax.scatter(df.index, df['H-L'], df['Volume'])

# Ajouter les étiquettes des axes X, Y et Z
ax.set_xlabel('Index')
ax.set_ylabel('H-L')
ax.set_zlabel('Volume')

# Afficher le nuage de points 3D
plt.show()
```
